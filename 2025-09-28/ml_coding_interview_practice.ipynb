# 10. Interview habits

While coding:

1. State expected tensor/dataframe shapes.
2. Start with the simplest correct solution.
3. Test small pieces before building the whole pipeline.
4. If something fails, inspect values/shapes rather than guessing.
5. Explain data leakage risks.
6. Use a small sanity-check batch before full training.
7. If loss is NaN, inspect:
   - learning rate
   - logits
   - input scale
   - gradients
   - invalid labels
8. If training does not improve, first verify that the model can overfit a tiny batch.
9. Never claim a metric is meaningful without understanding the split and class distribution.
10. Keep code readable. Interviewers care about your reasoning, not clever one-liners.
